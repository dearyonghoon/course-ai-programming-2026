## 7. Binary Classification Model 구성

````python
import pandas as pd
from google.colab import drive
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

drive.mount('/content/drive')

df = pd.read_csv(
    '/content/drive/MyDrive/Colab Notebooks/data/pima-indians-diabetes3.csv'
)

X = df.iloc[:, 0:8]  # column 0 - 7
y = df.iloc[:, 8]    # column 8

model = Sequential()
model.add(Input(shape=(8,)))
model.add(Dense(12, activation='relu', name='Dense_1'))
model.add(Dense(8, activation='relu', name='Dense_2'))
model.add(Dense(1, activation='sigmoid', name='Dense_3'))

model.summary()
````

### 모델 구조

```text
8 inputs
→ Dense(12, ReLU)
→ Dense(8, ReLU)
→ Dense(1, Sigmoid)
```

마지막 출력은 diabetes가 `1`일 확률을 나타냅니다.

## 8. Model Fitting

````python
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(
    X,
    y,
    epochs=100,
    batch_size=10
)
````

### 확인할 내용

- loss: `binary_crossentropy`
- optimizer: `adam`
- metric: `accuracy`
- epochs: `100`
- batch size: `10`

> 이번 실습의 목적은 **기본적인 binary classification 흐름을 경험하는 것**입니다.
