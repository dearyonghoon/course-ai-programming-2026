## 3. Data와 Label 준비

````python
import pandas as pd
from google.colab import drive

drive.mount('/content/drive')

df = pd.read_csv(
    '/content/drive/MyDrive/Colab Notebooks/data/iris3.csv'
)

X = df.iloc[:, 0:4]
y = df.iloc[:, 4]

y = pd.get_dummies(y, dtype=int)  # one-hot encoding for y

print(X.head(5))
print(y.head(5))
````

### One-Hot Encoding

예를 들어 class가 3개이면 label은 다음과 같이 표현됩니다.

```text
setosa      → [1, 0, 0]
versicolor  → [0, 1, 0]
virginica   → [0, 0, 1]
```

## 4. Multiclass Classification Model 구성

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential()
model.add(Input(shape=(4,)))
model.add(Dense(12, activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(3, activation='softmax'))

model.summary()
````

### 모델 구조

```text
4 inputs
→ Dense(12, ReLU)
→ Dense(8, ReLU)
→ Dense(3, Softmax)
```

마지막 3개의 출력은 각각 다음 class의 probability입니다.

```text
setosa
versicolor
virginica
```

## 5. Model Training

````python
model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(
    X,
    y,
    epochs=100,
    batch_size=10
)
````
